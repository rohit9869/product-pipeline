```sql
-- CELL 1
USE CATALOG product_ecommerce;
```

```sql

-- CELL 2
CREATE OR REPLACE TABLE product_ecommerce.silver.products
USING DELTA
AS SELECT
  product_id,
  product_name,
  category,
  sub_category,
  brand,
  CAST(price     AS DOUBLE)                                              AS price,
  CAST(cost      AS DOUBLE)                                              AS cost,
  CAST(stock_qty AS INT)                                                 AS stock_qty,
  CAST(rating    AS DOUBLE)                                              AS rating,
  ROUND(CAST(price AS DOUBLE) - CAST(cost AS DOUBLE), 2)                AS profit,
  ROUND(((CAST(price AS DOUBLE) - CAST(cost AS DOUBLE))
        / NULLIF(CAST(price AS DOUBLE), 0)) * 100, 2)                   AS profit_margin_percent,
  ROUND(CAST(price AS DOUBLE) * CAST(stock_qty AS INT), 2)              AS inventory_value,
  current_timestamp()                                                    AS processed_at
FROM product_ecommerce.bronze.products
WHERE product_id   IS NOT NULL
  AND product_name IS NOT NULL
  AND category     IS NOT NULL
  AND price        IS NOT NULL
  AND cost         IS NOT NULL
  AND stock_qty    IS NOT NULL
  AND CAST(price AS DOUBLE) > 0;

```

```sql
-- CELL 3
SELECT
  COUNT(*)                    AS total_records,
  COUNT(DISTINCT product_id)  AS unique_products,
  ROUND(AVG(price), 2)        AS avg_price,
  ROUND(AVG(profit), 2)       AS avg_profit,
  ROUND(AVG(rating), 2)       AS avg_rating
FROM product_ecommerce.silver.products;
```
