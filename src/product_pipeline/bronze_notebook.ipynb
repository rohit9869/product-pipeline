```sql
%python
file_path = "/Volumes/product_ecommerce/product/volume/products.csv"

spark.sql("USE CATALOG product_ecommerce")
spark.sql("CREATE SCHEMA IF NOT EXISTS bronze")
spark.sql("CREATE SCHEMA IF NOT EXISTS silver")
spark.sql("CREATE SCHEMA IF NOT EXISTS gold")

bronze_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(file_path)
)

bronze_df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("product_ecommerce.bronze.products")

print("Bronze loaded:", bronze_df.count(), "rows")
display(bronze_df)
```
