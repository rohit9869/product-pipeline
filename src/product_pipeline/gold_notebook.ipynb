```sql
-- CELL 1
USE CATALOG product_ecommerce;
```

```sql

-- CELL 2: Category Performance
CREATE OR REPLACE TABLE product_ecommerce.gold.category_performance
USING DELTA AS
SELECT
  category,
  COUNT(product_id)                     AS total_products,
  ROUND(AVG(price), 2)                  AS avg_price,
  ROUND(AVG(cost), 2)                   AS avg_cost,
  ROUND(AVG(profit), 2)                 AS avg_profit,
  ROUND(AVG(profit_margin_percent), 2)  AS avg_profit_margin_percent,
  SUM(stock_qty)                        AS total_stock,
  ROUND(SUM(inventory_value), 2)        AS total_inventory_value,
  ROUND(AVG(rating), 2)                 AS avg_rating
FROM product_ecommerce.silver.products
GROUP BY category
ORDER BY total_inventory_value DESC;
```

```sql

-- CELL 3: Brand Performance
CREATE OR REPLACE TABLE product_ecommerce.gold.brand_performance
USING DELTA AS
SELECT
  brand,
  COUNT(product_id)                     AS total_products,
  ROUND(AVG(price), 2)                  AS avg_price,
  ROUND(AVG(profit), 2)                 AS avg_profit,
  ROUND(AVG(profit_margin_percent), 2)  AS avg_profit_margin_percent,
  SUM(stock_qty)                        AS total_stock,
  ROUND(SUM(inventory_value), 2)        AS total_inventory_value,
  ROUND(AVG(rating), 2)                 AS avg_rating
FROM product_ecommerce.silver.products
GROUP BY brand
ORDER BY total_inventory_value DESC;
```

```sql

-- CELL 4: Top Profitable Products
CREATE OR REPLACE TABLE product_ecommerce.gold.top_profitable_products
USING DELTA AS
SELECT
  product_id,
  product_name,
  category,
  sub_category,
  brand,
  price,
  cost,
  profit,
  profit_margin_percent,
  stock_qty,
  inventory_value,
  rating
FROM product_ecommerce.silver.products
ORDER BY profit DESC
LIMIT 10;
```

```sql

-- CELL 5: Medallion Layer Count
CREATE OR REPLACE TABLE product_ecommerce.gold.medallion_layer_count
USING DELTA AS
SELECT 'Bronze - Raw Products'       AS layer, COUNT(*) AS records FROM product_ecommerce.bronze.products
UNION ALL
SELECT 'Silver - Cleaned Products'   AS layer, COUNT(*) AS records FROM product_ecommerce.silver.products
UNION ALL
SELECT 'Gold - Category Performance' AS layer, COUNT(*) AS records FROM product_ecommerce.gold.category_performance
UNION ALL
SELECT 'Gold - Brand Performance'    AS layer, COUNT(*) AS records FROM product_ecommerce.gold.brand_performance
UNION ALL
SELECT 'Gold - Top Products'         AS layer, COUNT(*) AS records FROM product_ecommerce.gold.top_profitable_products;
```

```sql

-- CELL 6: Final Checks
SELECT * FROM product_ecommerce.gold.category_performance;
SELECT * FROM product_ecommerce.gold.brand_performance;
SELECT * FROM product_ecommerce.gold.top_profitable_products;
SELECT * FROM product_ecommerce.gold.medallion_layer_count;
```
